```python
pontos = [hotel, cristo, jd_botanico, ccbb, museu_amanha, bondinho, aquario, theatro_mun,
          parque_patins, most_st_bento, lapa_st_teresa, forte_copa, zoo_rio, maracana, restaurante]
```

```python
nomes_pontos = {
    hotel: "Hotel",
    cristo: "Cristo Redentor",
    jd_botanico: "Jardim Botânico",
    ccbb: "CCBB",
    museu_amanha: "Museu do Amanhã",
    bondinho: "Pão de Açúcar",
    aquario: "AquaRio",
    theatro_mun: "Theatro Municipal",
    parque_patins: "Parque dos Patins",
    most_st_bento: "Mosteiro de São Bento",
    lapa_st_teresa: "Santa Teresa",
    forte_copa: "Forte de Copacabana",
    zoo_rio: "BioParque do Rio",
    maracana: "Maracanã",
    restaurante: "Restaurante Spesso"
}
```

```python
print(pontos)
print(len(pontos))  # Verifica o número de pontos na lista
```

```python
# Definindo um dicionário para armazenar as rotas com os nomes dos pontos
rotas_dict = {}

# Iterando sobre os pontos para criar as rotas
for i in range(len(pontos)):
    for j in range(len(pontos)):
        if i != j:
            origem = pontos[i]
            destino = pontos[j]
            
            # Usando os nomes dos pontos para facilitar a visualização
            nome_origem = nomes_pontos[origem]
            nome_destino = nomes_pontos[destino]
            
            # Consultando a API para obter os resultados de direção
            results = gmaps.directions(origin=origem, destination=destino, mode='walking')

            # Pegar o tempo total (em segundos) da primeira rota
            duration_seconds = results[0]['legs'][0]['duration']['value']

            # Converter para minutos
            duration_minutes = duration_seconds / 60
            if duration_minutes > 30:
                results = gmaps.directions(origin=origem, destination=destino, mode='transit', traffic_model='pessimistic')

            
            # Removendo as chaves indesejadas do resultado completo
            for route in results:
                # Remover chaves indesejadas no nível superior (não dentro de 'legs')
                for key in ['bounds','copyrights', 'overview_polyline', 'summary', 'warnings', 'waypoint_order']:
                    if key in route:
                        del route[key]

            # Salvando os resultados no dicionário com os nomes dos pontos como chave
            rotas_dict[(nome_origem, nome_destino)] = results
```

```python
# Exibindo o dicionário de rotas
for (origem, destino), rota in rotas_dict.items():
    print(f"Rota de {origem} para {destino}:")
    print(rota)  # ou qualquer outra informação que você deseje exibir
```

```python
# Criar uma lista para armazenar os dados extraídos
data = []

# Iterar sobre o dicionário de rotas
for (origem, destino), rotas in rotas_dict.items():
    for rota in rotas:
        # Inicializando as variáveis para cada rota
        total_distance_km = 0
        total_distance_m = 0
        total_duration_min = 0
        fare_currency = None
        fare_value = None
        step_modes = []

        # Verificar se o fare está presente
        if 'fare' in rota:
            fare_currency = rota['fare'].get('currency', None)
            fare_value = rota['fare'].get('text', None)

        # Iterar sobre as "legs" da rota
        for leg in rota.get('legs', []):
            total_distance_km += leg.get('distance', {}).get('value', 0) / 1000  # Convertendo metros para km
            total_distance_m += leg.get('distance', {}).get('value', 0)  # Metros
            total_duration_min += leg.get('duration', {}).get('value', 0) / 60  # Convertendo segundos para minutos

            # Capturar os modos de transporte de cada passo
            for step in leg.get('steps', []):
                travel_mode = step.get('travel_mode', 'UNKNOWN')

                # Identificar e formatar a descrição do modo de transporte
                if 'Bus towards' in step.get('html_instructions', ''):
                    step_modes.append('Bus')
                elif 'Subway towards' in step.get('html_instructions', ''):
                    step_modes.append('Subway')
                elif travel_mode == 'WALKING':
                    step_modes.append('Walking')
                elif travel_mode == 'BICYCLING':
                    step_modes.append('Bicycle')
                elif travel_mode == 'DRIVING':
                    step_modes.append('Driving')
                else:
                    step_modes.append(travel_mode)  # Caso algum outro modo seja identificado

        # Armazenar os dados na lista
        data.append({
            'origem': origem,
            'destino': destino,
            'total_distance_km': total_distance_km,
            'total_distance_m': total_distance_m,
            'total_duration_min': total_duration_min,
            'fare_currency': fare_currency,
            'fare_value': fare_value,
            'step_mode': ' | '.join(step_modes)  # Concatena os modos de transporte por |
        })

# Criar o DataFrame com os dados extraídos
df_rotas = pd.DataFrame(data)
```

```python
# Contar a ocorrência de cada modo de transporte em 'step_mode' para cada linha
df_rotas['walking'] = df_rotas['step_mode'].str.count('Walking')
df_rotas['bus'] = df_rotas['step_mode'].str.count('Bus')
df_rotas['subway'] = df_rotas['step_mode'].str.count('Subway')

df_rotas
```
